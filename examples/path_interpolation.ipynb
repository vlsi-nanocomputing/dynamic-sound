```python
import os
import dynamic_sound as ds
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

SMALL_SIZE = 12
MEDIUM_SIZE = 18
BIGGER_SIZE = 20

plt.rc('font', size=SMALL_SIZE)          # controls default text sizes
plt.rc('axes', titlesize=SMALL_SIZE)     # fontsize of the axes title
plt.rc('axes', labelsize=MEDIUM_SIZE)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('ytick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('legend', fontsize=SMALL_SIZE)    # legend fontsize
plt.rc('figure', titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
os.makedirs("_tmp", exist_ok=True)

src_folder = os.path.join("resources", "paths")
drone_path = ds.Path(file=os.path.join(src_folder, "drone_path_1sr.csv"))
microphone_path = ds.Path(file=os.path.join(src_folder, "camera_path_1sr.csv"))


# path interpolation
fig = plt.figure(figsize=(15, 10))
ax = drone_path.plot_path_3d(show=False, legend=False)
drone_path.plot_quaternion_directions(scale=1.0, show=False, ax=ax, legend=False)
microphone_path.plot_quaternion_directions(scale=1.0, show=False, ax=ax, legend=False)
ax.zaxis.set_major_formatter('{x:.0f}')
ax.xaxis.set_major_locator(MultipleLocator(5))
ax.yaxis.set_major_locator(MultipleLocator(5))
ax.set_aspect('equal')
ax.view_init(elev=30, azim=225)
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_zlabel("z (m)")
plt.tight_layout()
plt.savefig("_tmp/path_before_interpolation.png", dpi=300)
plt.show()

drone_path.interpolate_path(120)

fig = plt.figure(figsize=(15, 10))
ax = drone_path.plot_path_3d(show=False, legend=False)
drone_path.plot_quaternion_directions(scale=1.0, show=False, ax=ax, legend=False)
microphone_path.plot_quaternion_directions(scale=1.0, show=False, ax=ax, legend=False)
ax.zaxis.set_major_formatter('{x:.0f}')
ax.xaxis.set_major_locator(MultipleLocator(5))
ax.yaxis.set_major_locator(MultipleLocator(5))
ax.set_aspect('equal')
ax.view_init(elev=30, azim=225)
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_zlabel("z (m)")
plt.tight_layout()
plt.savefig("_tmp/path_after_interpolation.png", dpi=300)
plt.show()
```

```python
import dynamic_sound as ds
import os
import matplotlib.pyplot as plt
import soundfile as sf
import numpy as np
```

```python
os.makedirs("_tmp", exist_ok=True)
filename = os.path.join("_tmp", "sine_440Hz.wav")
sample_rate = 48000

signal, info = ds.generators.sine_generator(
    f=440,
    fs=sample_rate,
    amplitude=1.0,
    length="period",
    filename=filename,
    type=None
)
print(info)

signal = sf.read(filename)[0]
plt.plot(signal)
plt.show()
```

```python
os.makedirs("_tmp", exist_ok=True)
filename = os.path.join("_tmp", "white_noise.wav")
sample_rate = 48000

signal, info = ds.generators.noise_generator(
    fs=sample_rate,
    amplitude=1.0,
    length=12000,
    filename=filename,
    type=None
)
print(info)

signal = sf.read(filename)[0]
plt.plot(signal)
plt.show()
```
